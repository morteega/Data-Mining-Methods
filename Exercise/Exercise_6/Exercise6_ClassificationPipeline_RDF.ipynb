#### Read and check the data

```python
data = pd.read_csv('./data.csv',header='infer')


columns=data.columns
data[columns].isna().sum()
cat_cols=[]
for column in data:
    if data[column].dtype == 'O':
        cat_cols.append(column)

target=data["cancer_type"]
data=data.drop(columns="cancer_type")

num_vars=[]
for column in data:
    if data[column].dtype !="O":
        num_vars.append(column)
        
unnecesary_cols=["treatment_date", "diagnose_date"]

data=data.drop(columns=unnecesary_cols)
data.head(20)
data.dtypes

data["texture_2"]=pd.to_numeric(data["texture_2"], errors="coerce")

data.dtypes
data.head(20)

X_train_data, X_test_data, y_train, y_test=train_test_split(data,target, test_size=0.2, random_state=42)

for column in X_test_data:
    X_train_data[column]=X_train_data[column].fillna(X_train_data[column].mean())
    X_test_data[column]=X_test_data[column].fillna(X_train_data[column].mean())



model=RandomForestClassifier(n_estimators=30, max_depth=6,min_samples_leaf=2 ,random_state=1)

model=model.fit(X_train_data, y_train)

y_pred=model.predict(X_test_data)
y_train_result=model.predict(X_train_data)

print("Train accuracy:", accuracy_score(y_train, y_train_result))
print("Test accuracy:", accuracy_score(y_test, y_pred))

```
